## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

Lane: Structured Content Archetype Clustering.

Why: Instead of just looking at a binary "good vs. bad" performance, I want to group pages by how they actually behave in the real world to find recurring patterns (archetypes).

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The Search Question: What performance archetypes exist across the content inventory?

Unit of Analysis: A single, pseudonymized content item (page).

The Action: Based on the cluster a page falls into, a reviewer can confidently map it to an action like protect, improve, rewrite, merge, prune, or monitor.

Why Data/ML helps: Unsupervised learning (like K-Means clustering) can reveal hidden groupings across multiple dimensions (like volume, age, and engagement) that simple SQL threshold rules would miss.

Cost of a wrong call: Misclassifying a page could lead to accidentally pruning a "hidden gem" or wasting valuable editorial time rewriting a "stale visible page" that actually just needs a minor metadata update.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the starter anonymized dataset
df = pd.read_csv('content_refresh_anonymized.csv')

# Total inventory size
total_pages = len(df)

# Look at the variance in performance to justify clustering
active_pages = df[df['impressions_90d'] > 0]
median_impressions = active_pages['impressions_90d'].median()
max_impressions = active_pages['impressions_90d'].max()

print(f"1. Total inventory: {total_pages} pages available for clustering.")
print(f"2. Median 90-day impressions: {median_impressions}")
print(f"3. Max 90-day impressions: {max_impressions}")
print("\nWhy this justifies our lane:")
print("The massive gap between the median and max shows extreme variance in performance. Simple averages won't work to understand this inventory; we need archetypes (clustering) to group pages that behave similarly.")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*


What I can claim: I can claim to have discovered observed mathematical groupings (archetypes) based on historical metrics like search volume, age, and engagement. I can use these clusters to provide directional, decision-support recommendations to help reviewers prioritize which pages to look at first.

What I cannot claim: I cannot claim causal proof that editing a page will force it into a higher-performing cluster. I cannot claim to be "predicting Google's algorithm." Finally, because I am using performance metrics and metadata rather than raw article text, I cannot claim to be doing "semantic clustering."
