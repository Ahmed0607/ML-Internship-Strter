## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Lane: Structured Content Archetype Clustering.

ML Task Type: Unsupervised Learning (Clustering). We are not predicting a predefined outcome; instead, we are discovering hidden structural groupings (archetypes) across the content inventory.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target/Proxy: Because this is an unsupervised learning task, there is no explicit historical label to predict. The proxy for our target is the Cluster ID assigned by the algorithm (e.g., K-Means). This ID represents the archetype a specific page belongs to based on its metrics.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Technical Metric: Silhouette Score (to measure how well-separated and cohesive the clusters are) and Inertia.

Business Metric: Actionability and stability. The model is successful if the resulting archetypes map cleanly to specific content actions (e.g., protect, rewrite, merge, prune) and if the clusters remain stable when rerun on new data.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed SQL rule (e.g., IF impressions_90d < 100 AND content_age_days > 180 THEN 'prune') is rigid, relies on arbitrary human thresholds, and typically only evaluates 2 to 3 variables at a time. Machine Learning (clustering) beats this by evaluating pages across dozens of dimensions simultaneously, finding organic, mathematically backed performance patterns that a human writing rules would miss.
