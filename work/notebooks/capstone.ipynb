# Structured Content Archetype Clustering: Moving Beyond Heuristics

**Abstract:** This project identifies performance archetypes across a search inventory to improve editorial prioritization. Initially, a rigid heuristic was built to flag "Stale Visible Pages," but it struggled with nuance. To improve this, an unsupervised K-Means clustering model was developed to group pages by multi-dimensional behavioral patterns. However, strict client-grouped validation revealed severe generalization limits: while a naive random split yielded a 0.700 Precision@50, isolating unseen domains dropped precision to 0.220 (below the 0.440 base rate). This highlights that structural SEO metrics heavily overfit to specific domain architectures, proving that cross-domain content scoring requires semantic context, not just behavioral telemetry.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

**Data Scope:** This analysis utilizes the FlyRank/internship-warehouse release (v20260703). The grain is a single pseudonymized content item's aggregated performance for the mid-panel month of March 2026.
Exclusions: I deliberately excluded raw text and product-derived context flags (like health_score) to avoid circular logic and maintain public data safety, clustering strictly on observable search behavior.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

* **Baseline:** A multiplicative rule penalizing pages that were stale (180+ days), highly visible (500+ impressions), and slipping in position.
* **Model:** K-Means clustering ($k=4$) applied to scaled numerical features.
* **Validation Design:** An 80/20 Grouped Train/Test Split on client_id was strictly enforced. This was critical to prevent client-overlap leakage, ensuring the model didn't simply memorize the formatting or native traffic baselines of specific domains.

**Interpretation:** The model experienced a massive precision collapse when tested on completely unseen clients. Structural clustering successfully identifies decaying pages within a known domain, but struggles to generalize a universal "At-Risk" archetype across different websites without topical/semantic context.

## 5. Limitations

*What this work cannot claim.*

This is a directional, decision-support tool, not causal proof of Google's algorithm. The primary limitation is the "semantic blind spot." By excluding raw text, the model groups pages that behave similarly but cannot verify if they cover similar intents. The heavy reliance on impressions_90d also caused false positives, occasionally flagging high-traffic evergreen pages as "At-Risk".
